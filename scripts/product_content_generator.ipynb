## Set environment variables\n

import os

os.environ['GEMINI_API_KEY'] = ''
os.environ['MONGODB_URI'] = ''
os.environ['MONGODB_DB'] = 'allremotes'

```python
import os
import subprocess

result = subprocess.run(
    [
        'python', 'product_content_generator.py',
        '--mode', 'all',
        '--base-url', 'https://generativelanguage.googleapis.com/v1beta/openai/',
        '--model', 'gemini-3.1-flash-lite',
        '--api-key', os.environ.get('GEMINI_API_KEY', ''),
        '--uri', os.environ.get('MONGODB_URI', ''),
        '--db', 'allremotes',
        '--collection', 'products',
        '--log', 'product_content_google_log.jsonl',
        '--rpm', '5',
        '--tpm', '250000',
    ],
    capture_output=True,
    text=True,
)

print(result.stdout)
print(result.stderr)
```
